```python
import pandas as pd
```

```python
df = pd.read_csv("customer_shopping_behavior.csv")
df.head()
```

```python
df.info
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
# review rating ko 0 krne ne lia kyu ki kuch value NALL ho jati hai jisse replace krne me kabhi tym lgata hai 
# df["Review Rating"] = df.groupby("Payment Method")["Review Rating"].transform(lambda x : x.fillna(x.median()))

df["Review Rating"] = df["Review Rating"].fillna(df["Review Rating"].median())
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns=  df.columns.str.replace(' ', '_')
df.columns = df.columns.str.replace('purchase_amount_(usd)', 'purchase_amount_usd' )
```

```python
df.columns
```

```python
#  add a new column for age 
# labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
# df['age_group'] = pd.qcut(df['age'], q= 4, labels = labels)

df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 18, 30, 45, 70, 100],
    labels=['young', 'Adult', 'Middle-aged', 'Senior', '70+']
)
```

```python
df[['age','age_group']].head()
```

```python
print(df.shape)
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
# add a new column frequency mapping

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
