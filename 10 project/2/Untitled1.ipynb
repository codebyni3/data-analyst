```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
data =[]
```

```python
for page in range(1,10):
    url = f"https://quotes.toscrape.com/page/{page}/"
    result = requests.get(url)
    soup = BeautifulSoup(result.text, "html.parser")
    all_page = soup.find_all("div", {"class":"quote"})

    for page in all_page:
        # Quote = page.find("span",{"class":"text"})
        # Author = page.find("small",{"class":"author"})
        # tag = page.find_all("a",{"class":"tag"})
        Quote = page.find("span",class_ ="text").get_text(strip = True)
        Author = page.find("small",class_ ="author").get_text(strip = True)
        tag = page.find_all("a",class_ ="tag")
        for t in tag:
           if t.get_text(strip=True):
               # .text.strip()== "life":

                data.append({
                        "Quote":Quote,
                        "Author":Author,
                        "Tag" : "life"
                    })
print(data)
```

```python
df = pd.DataFrame(data)

print(df)
```
