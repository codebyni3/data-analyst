```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
books = []
```

```python
for page in range(1,30):
    url = f"https://books.toscrape.com/catalogue/page-{page}.html"
    r = requests.get(url)
    soup = BeautifulSoup(r.text,"html.parser")

    all_page = soup.find_all("article", {"class":"product_pod"})
    for page in all_page:
        title = page.h3.a["title"]
        rating = page.find("p",{"class":"star-rating One"})
        price = page.find("p",{"class":"price_color"})

        books.append({
            "Title":title,
            "Rating":rating,
            "price":price
           })
        
    
```

```python
df= pd.DataFrame(books)
df.to_csv("data1.csv",)
```
