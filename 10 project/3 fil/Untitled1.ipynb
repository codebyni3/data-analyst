```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
Product_name = []
Price = []
Description = []
Review = []
```

```python
for page in range(2,12):
    url = "https://www.flipkart.com/search?q=new+launch+tv&sid=ckf%2Cczl&as=on&as-show=on&otracker=AS_QueryStore_OrganicAutoSuggest_1_12_na_na_ps&otracker1=AS_QueryStore_OrganicAutoSuggest_1_12_na_na_ps&as-pos=1&as-type=RECENT&suggestionId=new+launch+tv%7CTelevisions&requestId=33ae24d5-65fa-4e7d-ae1b-36f74371f62d&as-backfill=on&otracker=nmenu_sub_TVs+%26+Appliances_0_New+Launches&page="+str(page)
    re = requests.get(url)
    soup = BeautifulSoup(re.text)
    box = soup.find("div", class_ = "QSCKDh dLgFEE")

    names = box.find_all("div",class_ = "RG5Slk")
    for i in names:
        name = i.text
        Product_name.append(name)
    
    price = box.find_all("div", class_ = "hZ3P6w DeU9vF")
    for i in price:
        name = i.text
        Price.append(name)

    des = box.find_all("div",class_ = "CMXw7N")
    for i in des:
        name = i.text
        Description.append(name)
    
    review = box.find_all("div", class_="a7saXW")
    for i in review:
        name = i.text
        Review.append(name)
```

```python
length = min(
    len(Product_name),
    len(Price),
    len(Description),
    len(Review)
)

df =  pd.DataFrame({
    "Product_name": Product_name[:length],
    "Price": Price[:length],
    "Description": Description[:length],
    "Review": Review[:length]
})

df.to_csv("NewMobile.csv")
```
