```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
Product_name =[]
Price = []
Description = []
Rating = []
```

```python
for i in range(2,12):
    url = "https://www.flipkart.com/search?q=under+20000+mobile&sid=tyy%2C4io&as=on&as-show=on&otracker=AS_QueryStore_OrganicAutoSuggest_2_6_na_na_na&otracker1=AS_QueryStore_OrganicAutoSuggest_2_6_na_na_na&as-pos=2&as-type=RECENT&suggestionId=under+20000+mobile%7CMobiles&requestId=09874242-951a-4c9c-972c-27a274fd447e&as-searchtext=under+&page="+str(i)
    re = requests.get(url)
    soup=BeautifulSoup(re.text)
    box = soup.find("div",class_="QSCKDh dLgFEE")

    names = box.find_all("div", class_ = "RG5Slk")
    # add name
    for i in names:
        name = i.text
        Product_name.append(name)
    # print(Product_name)

    # prlice
    price = box.find_all("div",class_="hZ3P6w DeU9vF")
    for i in price:
        name = i.text
        Price.append(name)
    # print(Price)

    
    # /description 
    des = box.find_all("div",class_="CMXw7N")
    for i in des:
        name = i.text
        Description.append(name)
    # print(Description)

    
    # /rating 
    rev = box.find_all("div",class_="a7saXW")
    for i in rev:
        name = i.text
        Rating.append(name)
    # print(Rating)
                                            

```

```python
print("Product:", len(Product_name))
print("Price:", len(Price))
print("Description:", len(Description))
print("Rating:", len(Rating))
```

```python
length = min(
    len(Product_name),
    len(Price),
    len(Description),
    len(Rating)
)

df = pd.DataFrame({
    "Product_name": Product_name[:length],
    "Price": Price[:length],
    "Description": Description[:length],
    "Rating": Rating[:length]
})

df.to_csv("flipkart_mobile.csv", index=False, encoding="utf-8-sig")

print("CSV file successfully created!")
print(df.head())
```
