```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.airbnb.co.in/s/New-Delhi/homes?place_id=ChIJLbZ-NFv9DDkRzk0gTkm3wlI&refinement_paths%5B%5D=%2Fhomes&checkin=2026-10-02&checkout=2026-10-04&date_picker_type=CALENDAR&search_type=HOMEPAGE_CAROUSEL_CLICK"
r= requests.get(url)
soup = BeautifulSoup(r.text)

name_head = []
Price_head =[]
rating_head =[]
 
name = soup.find_all("div",{"class":"t1jojoys atm_g3_1kw7nm4 atm_ks_15vqwwr atm_sq_1l2sidv atm_lo_l4hv9n atm_h3_7iojc0 atm_9s_cj1kg8 atm_6w_1e54zos atm_fy_1vgr820 atm_7l_hfv0h6 atm_cs_1mexzig atm_w4_1eetg7c atm_ks_zryt35__1rgatj2 dir dir-ltr"})
for n in name:
    a= n.text
    name_head.append(a)
        
Price = soup.find_all("span",{"class":"uesosmo atm_7l_8m3bwy atm_cs_80qqwn atm_rd_14k51in atm_rq_glywfm atm_cs_l3jtxx__1v156lz dir dir-ltr"})
for l in Price:
    p= l.text
    Price_head.append(p)
    
rating = soup.find_all("span",{"class":"r4a59j5 atm_h_1h6ojuz atm_9s_1txwivl atm_7l_hfv0h6 atm_84_1y44olf atm_mk_h2mmj6 atm_mj_glywfm dir dir-ltr"})
for k in rating:
    r= k.text
    rating_head.append(r)

    # desc = soup.find_alld("div",{"class":"n1v28t5c"})
        

    np = soup.find("a",{"aria-label":"Next"}).get("href")
    cpn= "https://www.airbnb.co.in"+np
    
    url = cpn
    r= requests.get(url)
    soup= BeautifulSoup(r.text)

df = pd.DataFrame({"name":name_head, "Price": Price_head, "rating": rating_head})
print(df)
```
