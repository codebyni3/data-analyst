```python
import pandas as pd
import requests
url = ("https://api.thedogapi.com/v1/breeds")
```

```python
result = requests.get(url)
```

```python
result.status_code
```

```python
data = result.json()
```

```python
df = pd.json_normalize(data)
```

```python
df
```

```python
columns = [
     "id",
     "name",
     "bred_for",
     "breed_group",
     "life_span",
     "temperament",
     "origin"
]
```

```python
df = df.reindex(columns=columns)
print(df.head(5))

df.to_csv("new", index=false )
```
