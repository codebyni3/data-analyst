```python
import requests
from bs4 import BeautifulSoup
import pandas as pd 

quote =[]
```

```python
for page in range(1, 10):

    # Page URL
    if page == 1:
        url = "https://quotes.toscrape.com/"
    else:
        url = f"https://quotes.toscrape.com/page/{page}/"
        respones = requests.get(url)
        soup = BeautifulSoup(respones.text, "html.parser")
        all_article = soup.find_all("div", class_ ="quote" )    
    for article in all_article:
        text = article.find("span", class_="text").get_text(strip =True)
        author = article.find("small", class_="author").get_text(strip =True)
        tags = article.find_all("a", class_="tag")
        tag_list = [tag.get_text(strip =True) for tag in tags]
        quote.append({
             "Quote": text,
             "Author": author,
             "Tags": ", ".join(tag_list)
              })
```

```python
df = pd.DataFrame(quote)
```

```python
df.to_csv("quotes.csv", index= False)
```
