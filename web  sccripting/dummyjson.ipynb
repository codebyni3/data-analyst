```python
import requests
import pandas as pd

url = "https://dummyjson.com/products"

result = requests.get(url)

data = result.json()

# Print products
for product in data["products"]:
    print("Title:", product["title"])
    print("Price:", product["price"])
    print("Category:", product["category"])
    print("Rating:", product["rating"])
    print()

# Convert to DataFrame
df = pd.DataFrame(data["products"])

# Save CSV
df.to_csv("dummyjson.csv", index=False)

print(df.head())
```

```python
url1 = "https://dummyjson.com/carts"

result1 = requests.get(url1)
data1 = result1.json()

for cart in data1["carts"]:
    print("Cart ID:", cart["id"])
    print("User ID:", cart["userId"])
    print("Total Products:", cart["totalProducts"])
    print("Total Quantity:", cart["totalQuantity"])
    print("Total:", cart["total"])
    print("Discounted Total:", cart["discountedTotal"])
    print()

df1= pd.DataFrame(data1["carts"])
df1.to_csv("dummyCart.csv", index=False)
print(df1.head())
```

```python
url2= "https://dummyjson.com/users"
result2 = requests.get(url2)
data2 = result2.json()

for user in data2["users"]:
    print("id:",  user["id"])
    print("username:",  user["username"])
    print("phone:",  user["phone"])
    print("age:", user["age"])
    print("gender:",  user["gender"])
    print("email:",  user["email"])
    print("password:",  user["password"])
    print("birthDate:",  user["birthDate"])
    print("address:",  user["address"])
    print()

df3 = pd.DataFrame(data2["users"])
df3.to_csv("users.csv", index= False)
print(df3.head())
```
