```python
import requests
# from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://openlibrary.org/search.json"
params = {
    "q": "python programming",
    "limit": 500
}

page = requests.get(url, params=params)
page.status_code
print(page)
# soup = BeautifulSoup(page.text, "html")
# print(soup)
book = []
data = page.json()
for data1 in data["docs"]:
    book.append({
        "Title":data1.get("title"),
        "Author": ",".join(data1.get("author_name",[])),
        "First_Publish_Year":data1.get("first_publish_year"),
        "ISBN":data1.get("isbn",[None])[0],
        "Publisher": data1.get("publisher", [None])[0],
        "Editions": data1.get("edition_count")

    })

df = pd.DataFrame(book)
df.to_csv("open_library_books.csv", index= False)
print(df.head())
    
```
